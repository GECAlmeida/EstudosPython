## Análise inicial


```python
# método .agg serve para agregar valores e realizar cálculos, .NamedAgg é um método de agrupar os dados em uma coluna com o nome realizando a função, como min, max e mean
dados.agg(min_renda = pd.NamedAgg('renda', 'min'),
          max_renda = pd.NamedAgg('renda', 'max'),
          media_renda = pd.NamedAgg('renda', 'mean'))
```

```python
# estatisticas descritivas, separar nas categorias da coluna default \ pula linha
'''
dados.groupby('default') \
  .agg(min_renda = pd.NamedAgg('renda', 'min'),
       max_renda = pd.NamedAgg('renda', 'max'),
       media_renda = pd.NamedAgg('renda', 'mean'))
'''
#sem pular linha
dados.groupby('default').agg(
    min_renda = pd.NamedAgg('renda', 'min'),
    max_renda = pd.NamedAgg('renda', 'max'),
    media_renda = pd.NamedAgg('renda', 'mean')
)
```
